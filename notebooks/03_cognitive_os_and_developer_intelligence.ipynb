## 1. MemGPT Hierarchy & Context Window Economics

Language models have finite context windows. Even with 1M+ token windows, **"needle in a haystack" degradation** causes LLMs to ignore facts buried in deep context.

GMS organizes memory into three distinct tiers:
1. **Working Memory (In-Context)**: Active task objectives, project brief, immediate constraints (`activeContext.md`).
2. **Episodic Ledger**: Time-series log of daily events, tool traces, and commit milestones.
3. **Semantic / Procedural Archive**: Deduplicated, verified facts and operational runbooks.

Context packets are sized according to strict token budgets:
* **Small (2,000 tokens)**: Lightweight status queries and mobile checks.
* **Standard (6,000 tokens)**: Routine feature development and pair programming.
* **Deep (15,000 tokens)**: High-assurance refactoring, architectural reviews, and debugging.

## 2. AST Code Symbol Intelligence & Pre-Edit Risk Audits

Before modifying code, human software engineers check for:
* What functions and classes live in this file?
* What other modules import this file?
* Have we encountered subtle concurrency bugs, gotchas, or policy restrictions on these exact symbols in the past?

GMS implements `mem_file_context`:
It parses the target Python file using Python's standard library `ast` module, extracts symbol definitions (`FunctionDef`, `ClassDef`) and imports, and queries memory for symbol-specific gotchas before editing!


````python
# Run AST Pre-Edit Risk Audit on a real repository file
target_file = "/Users/mihirdeshpande/projects/global-memory/tools/context_retrieval_planner.py"

report_json = memory_mcp_server.tool_mem_file_context({
    "file_path": target_file,
    "project": "global"
})

report = json.loads(report_json)
print(f"File Path: {report['file_path']}")
print(f"AST Analyzed: {report['ast_analyzed']}")
print(f"Pre-Edit Risk Level: {report['pre_edit_risk_level']}")
print(f"Top Extracted AST Symbols: {report['ast_top_symbols']}")
print(f"AST Module Dependencies: {report['ast_dependencies']}")
print(f"Relevant Memories Found: {report['relevant_memories_found']}")
````

## 3. Reflexion & Error Crystallization

When a test fails, a compiler throws an error, or a shell command crashes, naive agents repeatedly retry similar commands or hallucinate fixes.

In GMS, `mem_learn_error`:
1. Extracts the root error signature from the traceback.
2. Identifies the failing command.
3. Formulates a persistent procedural rule.
4. Stores the rule in `~/memories/lessons/` so future agents never repeat the mistake!


````python
# Simulate a compiler/test error trace
sample_traceback = """
tests/test_retrieval.py:42: in test_vector_dimension_mismatch
    res = planner.retrieve(root, query="concurrency", enable_indexed_vector=True, embedding_model="mismatch")
E   ValueError: Model mismatch: requested 'mismatch', found ['qwen3-embedding:0.6b']
"""

# Crystallize the error into a procedural rule
rule_resp = memory_mcp_server.tool_mem_learn_error({
    "error_trace": sample_traceback,
    "command": "pytest tests/test_retrieval.py",
    "resolution": "Check distinct models in gms_embeddings before calling indexed vector search.",
    "project": "global-memory",
    "agent": "notebook-agent"
})

print("Crystallized Procedural Rule:")
print(rule_resp)
````
